# Nested Dataclasses

```python
import dataclasses
from dataclasses import *
```

```python
@dataclasses.dataclass
class Permission:
    id: int
    name: str

@dataclasses.dataclass
class Group:
    id: int
    name: str
    permissions: list[Permission]

@dataclasses.dataclass
class User:
    id: int
    alias: str
    permissions: list[Permission]
    groups: list[Group]

def indent(n: int):
    print(n * '    ', end="")

    
def show_field(f, indent_level=0):
    for i, key in enumerate(["name", "type", "default", "default_factory"]):
        value = getattr(f, key, "<NA>")
        indent(indent_level)
        print(f"{'- ' if i==0 else '  '}{key}: {value!r}")

def show_dc(dc, indent_level=0):
    indent(indent_level)
    print(f"{dc.__name__}")
    for f in fields(dc):
        show_field(f)
```

```python
show_dc(User)
```
