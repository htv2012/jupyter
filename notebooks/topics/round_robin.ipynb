# Round Robin

```python
def round_robin(*iterables):
    i = 0
    iterators = [iter(x) for x in iterables]
    
    while iterators:
        try:
            out = next(iterators[i])
            i += 1
            yield out
        except StopIteration:
            del iterators[i]
        except IndexError:
            i = 0
```

```python
list(round_robin("abc", "de", "f"))
```

```python
list(round_robin("abc", "def", "gh"))
```

```python
list(round_robin("a", "def", "ghi"))
```
