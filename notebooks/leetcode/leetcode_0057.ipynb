```python
intervals = [[1,2],[3,3],[6,7],[8,10],[12,16]]
intervals
```

```python
newInterval = [4, 8]
newInterval
```

```python
ip = bisect.bisect_left(intervals, newInterval[0], key=lambda t: t[0])
print(f"{ip=}, at {intervals[ip]}")
```

```python
left = ip
while overlap(newInterval, intervals[left]):
    left -= 1
print(f"merge up to index {left} {intervals[left]}")
```
