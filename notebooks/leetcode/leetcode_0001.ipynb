```python
nums = [1, 2, 3, 0, 1, 5, 2]
indices = {}
for i, value in enumerate(nums):
    indices.setdefault(value, []).append(i)

print(f"{nums = }")
print(f"{indices = }")
```

```python
target = 2
for value, idx in indices.items():
    print(f"{value = }")
    if target == 2 * value and len(idx) >= 2:
        print(f"> Found {idx[:2]}")
        break
    elif target - value in indices:
        print(f"> found {idx[0]}, {indices[target - value][0]}")
        break
```

```python
nums = [1, 5, 7, 8]
target = 8

print(f"{target = }")
print(f"{nums = }")

for i, value in enumerate(nums):
    other = target - value
    other_i = bisect.bisect_left(nums, other, lo=i + 1)
    print(f"[{i}]{value}, [{other_i}]{other}")
    try:
        print(value + nums[other_i])
    except IndexError:
        pass
    if other_i < len(nums) and value + nums[other_i] == target:
        print(f"Found: {i + 1}, {other_i + 1}")
```
