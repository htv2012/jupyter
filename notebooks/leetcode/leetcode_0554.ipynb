```python
wall = [[1,2,2,1],[3,1,2],[1,3,2],[2,4],[3,1,2],[1,3,1,1]]
for row in wall:
    for brick in row:
        print(brick * "#", end="|")
    print()
```

```python
for row in wall:
    print(list(itertools.accumulate(row)))
```
