```python
from loguru import logger


def c(n):
    out = []
    k = 1

    while n != 0:
        if n % 2 == 1:
            out.append('1')
            n -= k
        else:
            out.append('0')
        n //= 2
        k *= -1

    out.reverse()
    return ''.join(out) or '0'
```

```python
c(2)
```

```python
c(3)
```

```python
c(4)
```
