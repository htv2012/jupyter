```python
import re, shlex
```

```python
class Fraction:
    def __init__(self, num, dem):
        self.num = int(num)
        self.dem = int(dem)

    def __repr__(self):
        return f"F({self.num}, {self.dem})"

    def __str__(self):
        return f"{self.num}/{self.dem}"
```

```python
ex = '1/2+3/10-5/8'
stack = []
tokens = shlex.shlex(ex)
for token in tokens:
    if token.isdigit() and len(stack) > 1 and stack[-2] == "/":
        stack.pop()
        num = stack.pop()
        stack.append(Fraction(num, token))
    else:
        stack.append(token)
stack        
```
