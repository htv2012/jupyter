# Leetcode 541 reverse string II

```python
import itertools
import logging
logging.basicConfig(level="DEBUG")
logger = logging.getLogger()
```

```python
s = "abcdefg"
k = 2

a, b = 0, 2 * k
buf = list(s)

while a < len(buf):
    chunk = buf[a:a+k]
    buf[a:a+k] = chunk[::-1]
    a, b = b, min(b + (2 * k), len(buf))

out = "".join(buf)
logger.debug(f"{buf=}")
logger.debug(f"{out=}")
```
