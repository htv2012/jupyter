# leetcode 1670 design a front middle back queue

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def __repr__(self):
        return f"({self.val})"
        

class FrontMiddleBackQueue:
    def __init__(self):
        self.front = None
        self.back = None
        self.mid = None
        self.len = 0
        
    def pushFront(self, val: int) -> None:
        node = Node(val)
        if self.front is None:
            self.front = self.back = self.mid = node
        else:
            node.right = self.front
            self.front = node
        self.len += 1
        
    def pushMiddle(self, val: int) -> None:
        pass

    def pushBack(self, val: int) -> None:
        self.back.append(val)

    def popFront(self) -> int:
        return self.front.pop()

    def popMiddle(self) -> int:
        pass

    def popBack(self) -> int:
        return self.back.pop()

    def __repr__(self):
        return f"<Q len: {self.len}, front: {self.front}, mid: {self.mid}, back: {self.back}>"

    def __iter__(self):
        node = self.front
        while node:
            yield node
            node = node.right

# Your FrontMiddleBackQueue object will be instantiated and called as such:
# obj = FrontMiddleBackQueue()
# obj.pushFront(val)
# obj.pushMiddle(val)
# obj.pushBack(val)
# param_4 = obj.popFront()
# param_5 = obj.popMiddle()
# param_6 = obj.popBack()
```
