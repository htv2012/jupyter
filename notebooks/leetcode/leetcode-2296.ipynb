# leetcode 2296 design a text editor

```python
import collections


class TextEditor:
    def __init__(self):
        self.pre = collections.deque()
        self.post = collections.deque()
        self.cursor = 0

    def __repr__(self):
        pre = "".join(self.pre)
        post = "".join(self.post)
        out = f"<{pre}|{post}>"
        return out

    def addText(self, text: str) -> None:
        self.pre.extend(text)

    def deleteText(self, k: int) -> int:
        count = min(k, len(self.pre))
        for _ in range(count):
            self.pre.pop()
        return count

    def cursorRight(self, k: int) -> str:
        count = min(k, len(self.post))
        for _ in range(count):
            self.pre.append(self.post.popleft())

        count = min(k, 10, len(self.pre))
        buf = []
        for i in range(-count, 0):
            buf.append(self.pre[i])
        return "".join(buf)
```
