# Leetcode 322 Coin Change

```python
import logging
from typing import List

logging.basicConfig(level="WARN")
logger = logging.getLogger()
logger.level = logging.WARN

class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        coins.sort(reverse=True)
        count = 0
        
        for coin in coins:
            if amount == 0:
                break
        
            number_of_coins, amount = divmod(amount, coin)
            logger.debug(f"{coin=}, {amount=}, {number_of_coins=}")
            count += number_of_coins
        logger.debug(f"{count=}, {amount=}")
        return count if amount == 0 else -1

sol = Solution()
f = sol.coinChange
```

```python
f([1,2,5], 11)
```
