# Binary Tree Sandbox

```python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.val = value
        self.left = left
        self.right = right

    def __repr__(self):
        left = f" L={self.left.val}" if self.left else ""
        right = f" R={self.right.val}" if self.right else ""
        return f"<TreeNode {self.val}{left}{right}>"
```

## Can we use enum to set left & right?

```python
import enum

class Side(str, enum.Enum):
    LEFT = "left"
    RIGHT = "right"

node1000 = TreeNode(1000)
node500 = TreeNode(500)
print(node1000)
print(node500)
```

```python
side = Side.LEFT
setattr(node1000, side, node500)
print(node1000)
print(node500)
```
